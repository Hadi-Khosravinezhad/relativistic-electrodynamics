```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# -------------------------------------------------------------
# 1. Trajectory Generation (Smooth & Slow Motion)
# -------------------------------------------------------------
# Phase 1: Straight line in field-free region (x < 0) - 45 frames
x_straight = np.linspace(-3.0, 0.0, 45)
y_straight = np.zeros_like(x_straight)

# Phase 2: Upward circular deflection in magnetic field (x >= 0) - 75 frames
R = 3.0  # Curvature radius
theta = np.linspace(0, np.pi / 3.2, 75)
x_curve = R * np.sin(theta)
y_curve = R * (1.0 - np.cos(theta))

# Total trajectory: 120 frames for very slow and smooth motion
x = np.concatenate([x_straight, x_curve])
y = np.concatenate([y_straight, y_curve])
n_frames = len(x)

# -------------------------------------------------------------
# 2. Figure Setup & Field Visualization
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(7, 4.5), dpi=100)

ax.set_xlim(-3.2, 2.8)
ax.set_ylim(-0.8, 1.8)
ax.set_aspect('equal')

ax.grid(True, linestyle='--', alpha=0.3)
ax.set_title("Upward Deflection of a Charged Particle in a Magnetic Field", fontsize=11, pad=12)
ax.set_xlabel("x", fontsize=10)
ax.set_ylabel("y", fontsize=10)

# Field boundary at x = 0
ax.axvline(0, color='#94a3b8', linestyle=':', lw=1.2, alpha=0.7)
ax.text(-0.1, 1.6, "Field Boundary", fontsize=8, color='#64748b', ha='right')

# Draw Magnetic Field in region x >= 0 (Pointing into the page: circled crosses)
gx, gy = np.meshgrid(np.linspace(0.4, 2.5, 5), np.linspace(-0.5, 1.5, 5))
ax.scatter(gx, gy, marker='o', s=130, facecolors='none', edgecolors='#0284c7', alpha=0.35, lw=1.2)
ax.scatter(gx, gy, marker='x', s=45, color='#0284c7', alpha=0.35, lw=1.2)
ax.text(1.45, 1.6, r"$\mathbf{B}$ Field ($\otimes$ into page)", fontsize=9, color='#0284c7', ha='center', weight='bold')

# Moving Elements: Trajectory trail & Charged particle
trail, = ax.plot([], [], color='#0284c7', lw=2.2, alpha=0.7)
particle, = ax.plot([], [], 'o', color='#ef4444', markersize=9)

# -------------------------------------------------------------
# 3. Animation Update Function
# -------------------------------------------------------------
def update(i):
    trail.set_data(x[:i+1], y[:i+1])
    particle.set_data([x[i]], [y[i]])
    return trail, particle

# Slow pacing: interval=65 ms, fps=15 (Total video duration ~ 8 seconds)
anim = FuncAnimation(fig, update, frames=n_frames, interval=65, blit=True)

# -------------------------------------------------------------
# 4. Save as MP4 Video & Render in Kaggle
# -------------------------------------------------------------
anim.save('particle_deflection.mp4', writer='ffmpeg', fps=15, extra_args=['-vcodec', 'libx264'])
plt.close(fig)

# Interactive HTML5 Video Player inside Kaggle
HTML(anim.to_html5_video())
```
